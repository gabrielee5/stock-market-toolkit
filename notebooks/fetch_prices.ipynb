# Fetch Prices

Run the price fetch step by step and visualize the result.
Set `ticker` and `timeframe` below.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / 'scripts'))

from fetch_prices import fetch_prices, clean_data, save_results
```

```python
ticker = 'AAPL'
timeframe = 'd'  # d | w | m

df = fetch_prices(ticker, timeframe)
print(f'{len(df)} rows from {df.index.min().date()} to {df.index.max().date()}')
df.tail()
```

```python
ax = df['Close'].plot(title=f'{ticker} ({timeframe}) - Close', figsize=(12, 5))
ax.set_ylabel('Close')
ax.grid(True, alpha=0.3)
```
